```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
```

```python
with open('StandardScaler.pkl', 'rb') as f:
    scaler=pickle.load(f)
with open('OneHotEncoding.pkl','rb') as f:
    ohe=pickle.load(f)
with open('LabelEncoder.pkl','rb') as f:
    label=pickle.load(f)

model=load_model('model.h5')
    
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
#now encoding the values
input_data['Gender'] = int(
    label.transform([input_data['Gender']])[0]
)
```

```python
list1=list(ohe.transform([[input_data['Geography']]]))
print(list1)
value=ohe.get_feature_names_out()
a=0
for i in value:
    input_data[i]=float(list1[0][a])
    a=a+1
del input_data["Geography"]
input_data
```

```python
df=pd.DataFrame([input_data])
```

```python
final=scaler.transform(df)
```

```python
prediction_proba=model.predict(final)
```

```python
if prediction_proba > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```

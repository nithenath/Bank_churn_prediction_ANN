```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
```

```python
training=pd.read_csv('datasets/training.csv')
testing=pd.read_csv('datasets/testing.csv')
x_train=training.drop(['Exited'],axis=1)
y_train=training['Exited']
x_test=testing.drop(['Exited'],axis=1)
y_test=testing['Exited']
```

```python
#creation of ANN model

model = Sequential([
    Dense(64, activation='relu', input_shape=(x_train.shape[1],)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
#creation of optimizers,loss functions 
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
#step up tensor board
tensorboard=TensorBoard(log_dir='logs',histogram_freq=1)
early_stopping=EarlyStopping(monitor="val_loss",patience=12,restore_best_weights=True)
```

```python
history = model.fit(
    x_train,
    y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=32,
    callbacks=[early_stopping, tensorboard],
    verbose=1
)
```
